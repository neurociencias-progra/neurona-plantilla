# 🟡 Etapa 4 — Issues, pruebas y tu primer worker

*📽 Vienes de GIT-02.*

**4a · Abre tus tres issues.** En la web, pestaña **Issues** de TU repositorio → **New
issue**, tres veces. Los títulos y cuerpos están en **`docs/issues-por-abrir.md`** (ábrelo

*📽 Ahora: TAL-02 — qué es una prueba, y el semáforo rojo → verde.*

**4b · Mira el rojo.** En la terminal (con `(.venv)`):

```
pytest
```

**Debes ver exactamente `4 failed, 8 passed`.** Las 8 verdes vigilan tu `TrenDeEspigas`
(¡ayer ya escribiste código probado!); las 4 rojas son tu lista de tareas de hoy. Elige UNA
roja y lee su fallo completo, de abajo hacia arriba: termina en la **pista** del
`NotImplementedError`.

**4c · Rellena el `Cargador`** (issue #1). Casi todo lo escribiste ya en los cuadernos de
**ficheros** (leer línea a línea) y de **tasa de disparo** (el CSV irregular y sus filas
fantasma): búscalo y tráelo — portar código honesto también es ingeniería.

* `metadatos()` → pista: `np.loadtxt` con `delimiter`, `usecols` y `skiprows`.
* `tiempos_de_espiga()` → ⚠️ línea vacía = ensayo VACÍO, **no se salta**: si la saltas, la
  fila *i* de espigas ya no corresponde a la fila *i* de metadatos y TODO lo demás queda
  mal **sin avisar** — el peor tipo de error.
* Itera: guarda → `pytest tests/test_cargador.py -q` → lee → corrige.

> 🤖 **Cameo de IA (así se usa bien):** pega tu *traceback* completo y pregunta *«explícame
> este error»* — no *«resuélveme el ejercicio»*. Todo uso se declara en `docs/reporte.md`.

Cuando creas que está, tu punto de control:

````python
# Punto de control · ¿Cargador listo? ✅
from neurona import Cargador

try:
    c = Cargador("datos/neu_400_RR034075_002_OCPSLA_10_00_VPCizq.csv")
    forma = c.metadatos().shape
    n_trenes = len(c.tiempos_de_espiga())
    print(f"metadatos: {forma} · trenes: {n_trenes}")
    if forma == (140, 31) and n_trenes == 140:
        print("✅ 140 ensayos, fila a fila. Corre `pytest` completo y ve a la Etapa 5.")
    else:
        print("🟡 Carga… pero los números no son los del laboratorio (¿saltaste líneas vacías?).")
except NotImplementedError as pista:
    print("⏳ Aún no. La pista:", pista)
````

# 🔵 Etapa 5 — El primer commit (el momento de la clase)

*📽 Vienes de GIT-03.*

**5a · Preséntate con git** — una sola vez por computadora, todos juntos. Usa **el mismo
correo de tu cuenta de GitHub** (para que tus commits salgan con tu foto):

```
git config --global user.name "Tu Nombre"
git config --global user.email "el-de-tu-cuenta@..."
```

**5b · El ciclo completo**, en la terminal:

```
git status
git add .
git status
git commit -m "Closes #1: el Cargador lee el CSV del laboratorio"
git push
```

⚠️ En el `push`, GitHub te pedirá autorizar **en el navegador** (una sola vez): di que sí
con tu sesión ya iniciada.

**5c · El momento 🤯:** recarga la pestaña **Issues** de tu repositorio en el navegador.
El **#1 debe aparecer CERRADO**, con tu commit enlazado — lo cerró el `Closes #1:` de tu
mensaje. Y para ver tu historia:

```
git log --oneline
```

**Debes ver** dos commits: el `Initial commit` y el tuyo. Tu proyecto ya tiene pasado. 📜
